```python
secret_number = 499999999999
max_attempts = 5000000000000
attempts_used = 0

while attempts_used < max_attempts:
    try:
        guess = int(input("Enter your guess: "))
    except ValueError:
        print("Invalid input. Please enter a whole number.")
        continue

    if guess < 0:
        print("Negative numbers are ignored. No attempt deducted.")
        continue

    attempts_used += 1

    if guess == secret_number:
        print("Congratulations! You guessed the secret number.")
        break
    else:
        remaining = max_attempts - attempts_used
        if remaining > 0:
            print(f"Wrong guess. Attempts remaining: {remaining}")
else:
    print("Game Over!")
```
